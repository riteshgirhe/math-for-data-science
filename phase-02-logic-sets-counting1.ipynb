## 1. Sets, Subsets, Relations and Functions

### Set
A **set** is an unordered collection of different elements.

Example: `A = {1, 2, 3}`

### Subset
`B` is a subset of `A` if every element of `B` is also in `A`.

Notation: `B ⊆ A`

### Relation
A relation connects elements from one set to another. It can be represented by ordered pairs such as `(x, y)`.

### Function
A function is a special relation where **every input has exactly one output**.

Example: `x → x²`

### Important set operations
- **Union:** everything in A or B
- **Intersection:** elements present in both A and B
- **Complement:** elements not in A

### Formula
`|A ∪ B| = |A| + |B| - |A ∩ B|`

### Example
A = `{1, 2, 3, 4}`  
B = `{3, 4, 5, 6}`

`A ∪ B = {1, 2, 3, 4, 5, 6}`

`A ∩ B = {3, 4}`

Therefore:

`|A ∪ B| = 4 + 4 - 2 = 6`



```python
# Set operations, plus a check against the inclusion-exclusion formula above

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

union = A | B
intersection = A & B
only_in_A = A - B

print("Union:", union)
print("Intersection:", intersection)
print("Only in A:", only_in_A)

# Verify |A ∪ B| = |A| + |B| - |A ∩ B|
formula_result = len(A) + len(B) - len(intersection)
assert len(union) == formula_result
print("\nInclusion-exclusion check passed:", len(union), "==", formula_result)
```

## 2. Propositional Logic, Induction and Contradiction

### Propositional logic
A proposition is a statement that is either **True** or **False**.

| Operator | Meaning |
|---|---|
| `A ∧ B` | AND |
| `A ∨ B` | OR |
| `¬A` | NOT |
| `A → B` | IMPLIES |

- **AND:** true only when both are true.
- **OR:** true when at least one is true.
- **NOT:** changes True to False and False to True.
- **IMPLIES:** false only when A is true and B is false.

### Proof by induction
Induction proves that a statement is true for all natural numbers.

Steps:
1. **Base case:** prove it for the first value.
2. **Inductive step:** assume it is true for `n`, then prove it for `n + 1`.

Think of it like dominoes: the first falls, and each falling domino knocks down the next one.

### Example
Prove:

`1 + 2 + 3 + ... + n = n(n + 1)/2`

For `n = 1`:

Left side = `1`

Right side = `1(2)/2 = 1`

So the base case is true.



```python
# Boolean logic in Python, matching the truth-table meanings above

A, B = True, False

print("A AND B:", A and B)   # ∧ — true only if both are true
print("A OR B :", A or B)    # ∨ — true if at least one is true
print("NOT A  :", not A)     # ¬ — flips the value
print("A -> B :", (not A) or B)  # → — implication is false only when A=True, B=False

# Quick check of the induction example: does the formula match the actual sum for several n?
def sum_formula(n):
    return n * (n + 1) // 2

for n in [1, 5, 10, 100]:
    actual = sum(range(1, n + 1))
    predicted = sum_formula(n)
    assert actual == predicted
    print(f"n={n}: actual sum={actual}, formula={predicted}  match: {actual == predicted}")
```

## 3. Counting Principles, Permutations and Combinations

This is the **important section** of Phase 2 for probability interviews.

### Counting principle
If one step has `m` choices and another step has `n` choices:

`Total ways = m × n`

Example: 1 letter and 2 digits:

`26 × 10 × 10 = 2600`

### Permutation
A permutation is an arrangement where **order matters**.

Formula:

`P(n, r) = n! / (n-r)!`

Example: assigning 1st, 2nd and 3rd place.

### Combination
A combination is a selection where **order does not matter**.

Formula:

`C(n, r) = n! / [r!(n-r)!]`

Example: choosing 3 people for a team.

### Easy way to remember
Ask:

**"Does changing the order change the answer?"**

- Yes → Permutation
- No → Combination

### Factorial
`n! = n × (n-1) × ... × 1`

And:

`0! = 1`

Example:

`5! = 120`

## Worked Examples

### Example 1 — Permutation
How many ways can 3 runners finish 1st, 2nd and 3rd out of 8?

Order matters.

`P(8,3) = 8! / 5! = 8 × 7 × 6 = 336`

### Example 2 — Combination
How many ways can we choose a 3-person committee from 8 people?

Order does not matter.

`C(8,3) = 8! / (3! × 5!) = 56`

### Example 3 — Repeated items
How many ways can we arrange `MISSISSIPPI`?

There are 11 letters:
- M = 1
- I = 4
- S = 4
- P = 2

Formula:

`11! / (1! × 4! × 4! × 2!)`

Answer:

`34,650`

```python
# Verify all three worked examples above by code

# Example 1
print("Example 1 - P(8,3):", permutation(8, 3), "-> expected 336")

# Example 2
print("Example 2 - C(8,3):", combination(8, 3), "-> expected 56")

# Example 3 - permutations with repeated letters
def permutations_with_repetition(n, repeat_counts):
    denom = 1
    for c in repeat_counts:
        denom *= factorial(c)
    return factorial(n) // denom

mississippi = permutations_with_repetition(11, [1, 4, 4, 2])  # M, I, S, P
print("Example 3 - MISSISSIPPI arrangements:", mississippi, "-> expected 34,650")
assert mississippi == 34650
```

## Counting and Probability

Counting is an important foundation for probability.

Example: probability of exactly 2 aces in a 5-card hand.

Choose:
- 2 aces from 4 → `C(4,2)`
- 3 other cards from 48 → `C(48,3)`
- Total 5-card hands → `C(52,5)`

So:

`P = [C(4,2) × C(48,3)] / C(52,5)`

The result is approximately `0.0399`, or about `4%`.

The main idea is: **count favorable outcomes and divide by total outcomes.**


```python
# Verify the poker-hand probability

favorable = combination(4, 2) * combination(48, 3)
total = combination(52, 5)
probability = favorable / total

print(f"Favorable hands: {favorable}")
print(f"Total 5-card hands: {total}")
print(f"P(exactly 2 aces) = {probability:.4f}  (~{probability*100:.2f}%)")
assert round(probability, 4) == 0.0399
```
