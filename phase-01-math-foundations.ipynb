## 1. Real Numbers and Complex Numbers

### Real Numbers

Real numbers are numbers that can be shown on a number line. They include:

- Natural numbers: 1, 2, 3, ...
- Whole numbers: 0, 1, 2, ...
- Integers: ..., -2, -1, 0, 1, 2, ...
- Rational numbers: fractions like 3/4 or 2/5
- Irrational numbers: numbers like √2 or π

**How they relate:**

```
Real Numbers
├── Rational
│   ├── Integers
│   │   ├── Whole numbers
│   │   └── Negative integers
│   └── Fractions / decimals
└── Irrational
    └── √2, √3, π, etc.
```

### Complex Numbers

Complex numbers go beyond real numbers.  
A complex number looks like this:

**z = a + bi**

- a = real part  
- b = imaginary part  
- i = imaginary unit (i² = -1)

**Example:**  
z = 3 + 4i  
- Real part = 3  
- Imaginary part = 4

### Why We Need Complex Numbers

Some equations have no real solution.  
Example: x² + 1 = 0  
→ x² = -1  

No real number squared gives -1.  
So we define i² = -1.  
Then the solutions are x = i and x = -i.

### Powers of i

- i¹ = i  
- i² = -1  
- i³ = -i  
- i⁴ = 1  

The pattern repeats every 4 powers: i, -1, -i, 1, ...

### Basic Operations

**Addition**  
(a + bi) + (c + di) = (a + c) + (b + d)i  

Example: (3 + 2i) + (4 + 5i) = 7 + 7i

**Subtraction**  
(a + bi) − (c + di) = (a − c) + (b − d)i  

Example: (3 + 2i) − (4 + 5i) = −1 − 3i

**Multiplication**  
Multiply normally and replace i² with −1.  

Example:  
(2 + 3i)(4 + i)  
= 8 + 2i + 12i + 3i²  
= 8 + 14i − 3  
= 5 + 14i

### Complex Conjugate

If z = a + bi,  
then its conjugate is z̄ = a − bi.

Example:  
z = 3 + 4i → z̄ = 3 − 4i

Important:  
z × z̄ = |z|²  

For z = 3 + 4i:  
3² + 4² = 25

### Magnitude (Modulus)

|z| = √(a² + b²)

Example:  
z = 3 + 4i  
|z| = √(9 + 16) = √25 = 5

This is the distance of the complex number from the origin (0,0).

### Complex Plane

- Horizontal axis = real part  
- Vertical axis = imaginary part  

So 3 + 4i is the point (3, 4).

## 2. Linear, Quadratic, Polynomial and Exponential Functions

| Function | Formula | Main idea |
|---|---|---|
| Linear | `y = mx + b` | Straight line |
| Quadratic | `y = ax² + bx + c` | Parabola |
| Polynomial | `a0 + a1x + ... + anx^n` | General polynomial |
| Exponential | `y = a*b^x` | Fast growth or decay |

### Important points

- Linear: Has a constant slope.
- Quadratic: Has a single vertex (minimum if a > 0 ( slope look like U ), maximum if a < 0 ( slope look like reverse U).
- Quadratic vertex: x = -b/(2a).
- Polynomial: Can have different degrees and shapes depending on its coefficients.
- Exponential: Changes by a constant factor rather than a constant difference.
- Exponential growth eventually becomes faster than polynomial growth.

### Example

At `x = 10`:

- `y = x` → `10`
- `y = x²` → `100`
- `y = 2^x` → `1024`

This idea is useful when thinking about growth rates and ML algorithms.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

plt.plot(x, x, label="Linear: x")
plt.plot(x, x**2, label="Quadratic: x²")
plt.plot(x, 2**x, label="Exponential: 2ˣ")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
````

## 3. Equations and Inequalities

### Linear equation

A linear equation has the variable to the first power.

Example:

`2x + 3 = 11`

Answer:

`x = 4`

### Quadratic equation

General form:

`ax² + bx + c = 0`

- **Quadratic form:** add a ≠ 0. Without it, the equation isn't quadratic and the formula divides by zero.

Quadratic formula:

`x = (-b ± sqrt(b² - 4ac)) / (2a)`

The value:

`Δ = b² - 4ac`

is called the **discriminant**.

- `Δ > 0` → two real roots
- `Δ = 0` → one repeated real root
- `Δ < 0` → two complex roots
- ***Discriminant, Δ < 0:*** say "two complex conjugate roots (for real a, b, c)". Complex roots come in conjugate pairs, which ties back to topic 1.
  


### Inequality

Example:

`-3x + 6 > 0`

`-3x > -6`

Because we divide by a negative number, the sign changes:

`x < 2` also write like `it as an interval, (-∞, 2)`

**Key rule:** When multiplying or dividing an inequality by a negative number, reverse the inequality sign.

````python
import sympy as sp

x = sp.symbols('x')

# Linear equation
print(sp.solve(2*x + 3 - 11, x))

# Quadratic equation
print(sp.solve(x**2 - 5*x + 6, x))

# Inequality
print(sp.solve_univariate_inequality(-3*x + 6 > 0, x))
````

## 4. Cartesian and Polar Graphs

### Cartesian coordinates

A point is written as:

`(x, y)`

This is the normal coordinate system used in most charts.

### Polar coordinates

A point is written as:

`(r, θ)`

Here:
- `r` = distance from the origin
- `θ` = angle

### Conversion formulas

`x = r cos(θ)`

`y = r sin(θ)`

`r = sqrt(x² + y²)`

### Example

Convert `(3, 4)` to polar form.

`r = sqrt(3² + 4²) = 5`

`θ = atan2(y, x) ≈ 53.13°`

So the answer is approximately:

`(5, 53.13°)`

### Conic sections

The main types are:
- Circle
- Ellipse
- Parabola
- Hyperbola

These are mostly background knowledge for Data Science

````python
import numpy as np
y=4
x=3

np.atan2(y, x)                  # This is in radians
np.degrees(np.atan2(4, 3))      # This is in degree
````

## 5. Sequences, Series and Sums

### Sequence

A sequence is an ordered list of numbers.

Example:

`1, 2, 4, 8, 16, ...`

### Series

A series is the sum of sequence terms.

Example:

`1 + 2 + 4 + 8 + 16`

### Arithmetic sequence

The difference between terms is constant.

Example:

`3, 7, 11, 15, ...`

Common difference = `4`

Arithmetic sum:

`S_n = n/2 * (first term + last term)`

### Geometric sequence

The ratio between terms is constant.

Example:

`1, 2, 4, 8, ...`

Common ratio = `2`

Finite geometric sum:

`S_n = a(1-r^n)/(1-r)`

If `r = 1`, every term equals `a`, so `S_n = n·a`.

Infinite geometric sum, when `|r| < 1`:

`S = a/(1-r)`

### Example

`1 + 1/2 + 1/4 + 1/8 + ...`

Here `a = 1` and `r = 1/2`.

So:

`S = 1/(1-1/2) = 2`

This idea is useful later for Taylor series and probability distributions

## 6. Feature Scaling, Data Transformations and Activation Functions

This is the **most important section in Phase 1** for ML interviews.

### Feature scaling

Feature scaling changes numeric features to a similar scale.

Why?

Suppose:
- Age is between `0` and `100`
- Salary is between `0` and `1,000,000`

Without scaling, the large values can dominate distance calculations and gradient updates.

### Min-max scaling

Definition: Min-Max scaling transforms numerical values to a fixed range, usually 0 to 1.
Formula:

`x' = (x - min(x)) / (max(x) - min(x))`

Usually gives values between `0` and `1`

Example: Suppose ages are: 20, 30, 40, 50
For x = 30:

x' = (30 - 20) / (50 - 20)
`x' = 0.33 `So, 30 becomes approximately 0.33.

Note: Min-Max scaling is sensitive to outliers because the minimum and maximum determine the scaling range

### Z-score standardization

Definition: Z-score standardization transforms values based on how far they are from the mean, measured in standard deviations
Formula:

`x' = (x - μ) / σ`

Example: Suppose: Mean = 50,Standard deviation = 10,Value = 70
Then:
z = (70 - 50) / 10
`z = 2` So, 70 is 2 standard deviations above the mean.

The result has mean close to `0` and standard deviation close to `1`.

### Log transformation

A log transformation can reduce right skew.

Formula:

`x' = log(x)`

### Activation Functions

Activation functions add non-linearity to neural networks.

`Sigmoid`

$$ \sigma(x)=\frac{1}{1+e^{-x}} $$

Output: 0 to 1

`ReLU`

$$ f(x)=\max(0,x) $$

Positive → same value
Negative → 0

`Tanh`

Output: -1 to 1

`Softmax`

$$ Softmax(x_i)=\frac{e^{x_i}}{\sum_j e^{x_j}} $$

Converts outputs into probabilities that sum to 1. Commonly used for multi-class classification.

| Function       | Formula                            | Output          | Main use                          |
| -------------- | ---------------------------------- | --------------- | --------------------------------- |
| **Sigmoid**    | \(\frac{1}{1+e^{-x}}\)             | 0 to 1          | Binary classification             |
| **Tanh**       | \(\tanh(x)\)                       | -1 to 1         | Hidden layers / older RNNs        |
| **ReLU**       | \(\max(0,x)\)                      | 0 to ∞          | Hidden layers                     |
| **Leaky ReLU** | \(\max(0.01x,x)\)                  | -∞ to ∞         | Hidden layers; reduces dying ReLU |
| **Softmax**    | \(\frac{e^{x_i}}{\sum_j e^{x_j}}\) | 0 to 1, sum = 1 | Multi-class classification        |




````python
# Min-max scaling from scratch

def min_max_scale(x, min_value, max_value):
    return (x - min_value) / (max_value - min_value)

data = [10, 20, 30, 40, 50]

scaled = [min_max_scale(x, min(data), max(data)) for x in data]
scaled
````

````python
# Z-score standardization from scratch

import numpy as np

def z_score(x, mean, std):
    return (x - mean) / std

data = np.array([10, 20, 30, 40, 50])

mean = data.mean()
std = data.std()

standardized = z_score(data, mean, std)
standardized
````

````python
# Sigmoid and ReLU from scratch

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

values = np.array([-3, -2, 0, 2, 5])

print("Sigmoid:", sigmoid(values))
print("ReLU:", relu(values))
````
